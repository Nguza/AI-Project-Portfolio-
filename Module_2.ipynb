**Clean the dataset:**

Remove duplicates.

Handle missing values.

Standardize formats.


```python
df.duplicated().sum()
```

```python
#In case there are duplicates-optional
df.drop_duplicates(inplace=True)
```

```python
#check missing values in dataset
df.isnull().sum()
```

**Descriptive Analysis**

Compute Mean, Median, Mode, Std Dev

```python
print("Mean Sale Price:", df['price'].mean())
print("Median Sale Price:", df['price'].median())
print("Mode Sale Price:", df['price'].mode()[0])
print("Std Dev Sale Price:", df['price'].std())
```

#check skewness

```python
print("Skewness:", df['price'].skew())
```

```python
# Compute skewness
skewness = df['price'].skew()
print("Skewness:", skewness)

# Plot distribution with histogram and KDE (Kernel Density Estimate)
plt.figure(figsize=(10,6))
sns.histplot(df['price'], bins=30, kde=True, color='skyblue', edgecolor='black')
plt.title(f'Distribution of Sale Prices (Skewness = {skewness:.2f})')
plt.xlabel('Sale Price')
plt.ylabel('Frequency')
plt.show()
```

**Identify Outliers**

Can use IQR method or boxplot

```python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df['price'] < Q1 - 1.5*IQR) | (df['price'] > Q3 + 1.5*IQR)]
print("Number of outliers:", len(outliers))
```

```python
# Select numeric columns only
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Compute correlation
corr = numeric_df.corr()

# Get top 10 features most correlated with SalePrice
top_features = corr['price'].sort_values(ascending=False).head(10).index
top_corr = corr.loc[top_features, top_features]

# Plot heatmap
plt.figure(figsize=(10,8))
sns.heatmap(top_corr, annot=True, fmt=".2f", cmap="YlGnBu", cbar=True)
plt.title("Top 10 Features Correlation with SalePrice")
plt.show()
```
